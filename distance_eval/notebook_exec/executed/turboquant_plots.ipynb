# Turboquant — TQMSE + TQProd

Loads `*_{METHOD}_adc_vs_exact_eval.csv` from `/mnthdd/cpanourg/2-hdvc/results/urania_results/results/turboquant` for **both** `TQMSE` (`TQ-MSE-*`) and `TQProd` (`TQ-Prod-*`). See `turboquant/export_results_for_figures.py`.

Plot style matches `relerr_cpp_plots.ipynb` via `relerr_quick_plot_style.py`.

Outputs:
- `/mnthdd/cpanourg/2-hdvc/results/urania_results/results/turboquant/figures/TQMSE/`
- `/mnthdd/cpanourg/2-hdvc/results/urania_results/results/turboquant/figures/TQProd/`

**With RaBitQ on the same axes:** run `turboquant_rabitq_combined_plots.ipynb` (figures under `/mnthdd/cpanourg/2-hdvc/results/urania_results/results/figures/turboquant_rabitq_combined/`).

```python
import sys
from pathlib import Path

import pandas as pd

REPO = Path("/home/cpanourg/projects/2-hdvc")
sys.path.insert(0, str(REPO / "scripts" / "figure_generators"))

from relerr_quick_plot_style import figure_relerr_vs_nbits, savefig_relerr

DATA_DIR = Path("/mnthdd/cpanourg/2-hdvc/results/urania_results/results/turboquant")
METHODS = ["TQMSE", "TQProd"]

for METHOD in METHODS:
    out_dir = DATA_DIR / "figures" / METHOD
    paths = sorted(DATA_DIR.glob(f"*_{METHOD}_adc_vs_exact_eval.csv"))
    if not paths:
        print(f"Skip {METHOD}: no CSVs (run turboquant/export_results_for_figures.py)")
        continue
    for p in paths:
        df = pd.read_csv(p)
        ds = str(df["dataset"].iloc[0])
        fig, _ax = figure_relerr_vs_nbits(df, method_label=METHOD)
        out = out_dir / f"{ds}_{METHOD}_relerr_vs_nbits.pdf"
        savefig_relerr(fig, out)
        print(out)
```
