# Turboquant + RaBitQ — combined relative error vs nbits

For each dataset, one figure with **TQMSE**, **TQProd**, and **RaBitQ** (same relerr style as `relerr_cpp_plots.ipynb`).

Data:
- Turboquant: `/data/cpanourg/2-hdvc/results/turboquant/*_{TQMSE,TQProd}_adc_vs_exact_eval.csv`
- RaBitQ: `/data/cpanourg/2-hdvc/results/rabitq/*_RaBitQ_adc_vs_exact_eval.csv`

A dataset is skipped if no CSV exists for it. Curves are only added when the corresponding file is present.

```python
import sys
from pathlib import Path

import pandas as pd

REPO = Path("/home/cpanourg/projects/2-hdvc")
sys.path.insert(0, str(REPO / "scripts" / "figure_generators"))

from relerr_quick_plot_style import figure_relerr_vs_nbits_multi, savefig_relerr

TURBO_DIR = Path("/data/cpanourg/2-hdvc/results/turboquant")
RABITQ_DIR = Path("/data/cpanourg/2-hdvc/results/rabitq")
OUT = Path("/data/cpanourg/2-hdvc/results/figures/turboquant_rabitq_combined")

turbo_datasets = set()
for p in TURBO_DIR.glob("*_TQMSE_adc_vs_exact_eval.csv"):
    turbo_datasets.add(p.name.split("_")[0])
rabitq_datasets = {p.name.split("_")[0] for p in RABITQ_DIR.glob("*_RaBitQ_adc_vs_exact_eval.csv")}

datasets = sorted(turbo_datasets & rabitq_datasets)
print("Datasets with both turboquant + rabitq:", datasets)
only_turbo = sorted(turbo_datasets - rabitq_datasets)
only_rq = sorted(rabitq_datasets - turbo_datasets)
if only_turbo:
    print("Turboquant only (no RaBitQ file):", only_turbo)
if only_rq:
    print("RaBitQ only (no turboquant files):", only_rq)

CURVES = [
    (TURBO_DIR / "{ds}_TQMSE_adc_vs_exact_eval.csv", "TQMSE"),
    (TURBO_DIR / "{ds}_TQProd_adc_vs_exact_eval.csv", "TQProd"),
    (RABITQ_DIR / "{ds}_RaBitQ_adc_vs_exact_eval.csv", "RaBitQ"),
]

for ds in datasets:
    series = []
    for path_tpl, label in CURVES:
        path = Path(str(path_tpl).format(ds=ds))
        if path.is_file():
            series.append((pd.read_csv(path), label))
    if not series:
        continue
    fig, _ax = figure_relerr_vs_nbits_multi(series, title=ds)
    out = OUT / f"{ds}_turboquant_rabitq_relerr_vs_nbits.pdf"
    savefig_relerr(fig, out)
    print(out)
```
