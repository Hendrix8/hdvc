# RaBitQ — relative error vs nbits

Loads `*_RaBitQ_adc_vs_exact_eval.csv` from `/data/cpanourg/2-hdvc/results/rabitq`.

Plot style matches `relerr_cpp_plots.ipynb` via `relerr_quick_plot_style.py` (same fonts, y-grid, markers, legend outside for nbits x-axis).

**Turboquant + RaBitQ on one figure per dataset:** see `turboquant_rabitq_combined_plots.ipynb` (TQMSE, TQProd, RaBitQ curves together).

```python
import sys
from pathlib import Path

import pandas as pd

REPO = Path("/home/cpanourg/projects/2-hdvc")
sys.path.insert(0, str(REPO / "scripts" / "figure_generators"))

from relerr_quick_plot_style import figure_relerr_vs_nbits, savefig_relerr

DATA_DIR = Path("/data/cpanourg/2-hdvc/results/rabitq")
OUT = DATA_DIR / "figures" / "RaBitQ"

paths = sorted(DATA_DIR.glob("*_RaBitQ_adc_vs_exact_eval.csv"))
assert paths, f"No RaBitQ CSVs under {DATA_DIR}"

for p in paths:
    df = pd.read_csv(p)
    ds = str(df["dataset"].iloc[0])
    method = str(df["method"].iloc[0])
    fig, _ax = figure_relerr_vs_nbits(df, method_label=method)
    out = OUT / f"{ds}_{method}_relerr_vs_nbits.pdf"
    savefig_relerr(fig, out)
    print(out)
```
